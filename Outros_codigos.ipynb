```python
import pandas as pd

# Criando o dicionário com os dados do exemplo
dados = {
    'ID_Pedido':[1,2,3,4,5,6],
    'Data': ['2026-01-10', '2026-01-11', '2026-01-12', '2026-01-13', '2026-01-14', '2026-01-15'],
    'Produto': ['Notebook', 'Smartphone', 'Cadeira Gamer', 'Monitor 24"', 'Mesa Escritório', 'Teclado Mecânico'],
    'Categoria': ['Eletrônicos', 'Eletrônicos', 'Móveis', 'Eletrônicos', 'Móveis', 'Acessórios'],
    'Quantidade':[1.0,2.0,2.0,2.0,2.0,1.0],
    'Valor_Unitario': [3500.00, 1200.00, 850.00, 900.00, 450.00, 250.00],
    'Regiao': ['Sudeste', 'Nordeste', 'Sul', 'Sudeste', 'Centro-Oeste', 'Norte']
}

# Transformando em um DataFrame do Pandas
df = pd.DataFrame(dados)

# Convertendo a coluna Data para o tipo datetime correto
df['Data'] = pd.to_datetime(df['Data'])

# Salvando o arquivo em formato Excel (.xlsx)
df.to_excel('dados_vendas.xlsx', index=False)

print("Arquivo 'dados_vendas.xlsx' criado com sucesso!")
```
